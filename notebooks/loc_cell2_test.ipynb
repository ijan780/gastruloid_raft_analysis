```python
import tifffile as tiff
from skimage import io
from skimage.exposure import rescale_intensity, equalize_adapthist
from skimage.morphology import disk, closing, opening, remove_small_holes, remove_small_objects
from skimage.filters import threshold_otsu, gaussian
import numpy as np

img_path = r'C:\Users\MVX\Documents\2 - Data\240805_R2GLR_1.8e6_0-48hrBMP4_0%aneu_4_Analysis\Individual Raft Images\Endpoint\array1_raft207_end.tif'
with tiff.TiffFile(img_path) as tif:
    img = tif.pages[0].asarray()

img = equalize_adapthist(img) 
# io.imshow(img)

img = gaussian(img, sigma=2)

threshVal = threshold_otsu(img)
thresh = img > threshVal
print(threshVal)

thresh = opening(thresh, disk(5))
thresh = remove_small_holes(thresh, area_threshold=5e4)
thresh = remove_small_objects(thresh, min_size=7.5e4)

io.imshow(thresh)
```
